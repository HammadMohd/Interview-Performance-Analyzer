# Preprocessing
Basic data understanding: shape, dtypes, nulls, duplicates, and summary statistics.

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('../data/features/merged_all.csv')
print('Shape:', df.shape)
df.head()
```

## 1. Data Types

```python
print(df.dtypes)
```

## 2. Null Values

```python
null_sum = df.isnull().sum()
null_pct = (df.isnull().sum() / len(df) * 100).round(2)
null_df = pd.DataFrame({'null_count': null_sum, 'null_pct': null_pct})
null_df = null_df[null_df['null_count'] > 0].sort_values('null_pct', ascending=False)
print('Columns with null values:')
print(null_df)
print(f'\nTotal null columns: {len(null_df)}')
```

## 3. Duplicates

```python
print('Duplicate rows:', df.duplicated().sum())
print('Duplicate IDs:', df['id'].duplicated().sum())
```

## 4. Numeric Summary

```python
df.describe()
```

## 5. User Distribution

```python
print('Unique users:', df['user_no'].nunique())
print('\nQuestions per user:')
print(df.groupby('user_no').size().value_counts().sort_index())
```
